```python
import pandas as pd
stocks=pd.read_csv('/Users/aja/Desktop/uni/datagath/stocks.csv',sep=';')

stocks['capital_gain_loss']=(-stocks['price']).where(stocks['operation']=='Buy',stocks['price'])

res = stocks.groupby('stock_name')['capital_gain_loss'].sum().reset_index()
res.sort_values(by='capital_gain_loss')
print(res)
```
