```python
import pandas as pd
users=pd.read_csv('/Users/aja/Desktop/uni/datagath/users.csv',sep=';')
movies=pd.read_csv('/Users/aja/Desktop/uni/datagath/movies2.csv',sep=';')
movie_rating=pd.read_csv('/Users/aja/Desktop/uni/datagath/movie_rating.csv',sep=';')

ans = movie_rating.merge(movies, on='movie_id', how='left')
ans = ans.merge(users, on='user_id', how='left')

ans['created_at'] = pd.to_datetime(ans['created_at'], format='%d.%m.%Y')
clean = ans[ans['created_at'].between('01.02.2020', '29.02.2020')]
clean = ans[(ans['created_at'] >= pd.to_datetime('01.02.2020', format='%d.%m.%Y')) & 
            (ans['created_at'] <= pd.to_datetime('29.02.2020', format='%d.%m.%Y'))]
clean.head()


user_count = movie_rating.groupby('user_id').size().reset_index(name='count')
user_count = user_count.merge(users, on='user_id', how='left')

top_user = user_count.sort_values(by=['count', 'name'], ascending=[False, True]).iloc[0]['name']



feb_avg_rating = clean.groupby('movie_id')['rating'].mean().reset_index(name='avg_rating')
feb_avg_rating.head()
feb_avg_rating = feb_avg_rating.merge(movies, on='movie_id', how='left')
feb_avg_rating.head()
top_movie = feb_avg_rating.sort_values(by=['avg_rating', 'title'], ascending=[False, True]).iloc[0]['title']





result = pd.DataFrame({'results': [top_user, top_movie]})

print(result)
```
